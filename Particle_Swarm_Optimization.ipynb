```python

import random
import math

def distance(city1, city2):
    return math.sqrt(
        (city1[0] - city2[0]) ** 2 +
        (city1[1] - city2[1]) ** 2
    )

def calculate_distance(route, cities):
    total = 0

    for i in range(len(route)):
        city1 = cities[route[i]]
        city2 = cities[route[(i + 1) % len(route)]]

        total += distance(city1, city2)

    return total


def pso_tsp(cities, particles=30, iterations=200):

    n = len(cities)


    swarm = []

    for i in range(particles):
        route = list(range(n))
        random.shuffle(route)
        swarm.append(route)


    pbest = [route[:] for route in swarm]

    pbest_cost = [
        calculate_distance(route, cities)
        for route in swarm
    ]


    best_index = pbest_cost.index(min(pbest_cost))
    gbest = pbest[best_index][:]

    for iteration in range(iterations):

        for i in range(particles):

            current = swarm[i]


            new_route = current[:]

            if random.random() < 0.8:
                a, b = random.sample(range(n), 2)
                new_route[a], new_route[b] = (
                    new_route[b], new_route[a]
                )


            if random.random() < 0.7:
                a, b = random.sample(range(n), 2)

                city = gbest[a]

                if city in new_route:
                    pos = new_route.index(city)
                    new_route[a], new_route[pos] = (
                        new_route[pos], new_route[a]
                    )

            swarm[i] = new_route

            cost = calculate_distance(new_route, cities)

            if cost < pbest_cost[i]:
                pbest[i] = new_route[:]
                pbest_cost[i] = cost

        best_index = pbest_cost.index(min(pbest_cost))

        if pbest_cost[best_index] < calculate_distance(gbest, cities):
            gbest = pbest[best_index][:]

        if (iteration + 1) % 50 == 0:
            print("Iteration:", iteration + 1,
                  "Best Distance:",
                  round(calculate_distance(gbest, cities), 2))

    return gbest, calculate_distance(gbest, cities)


cities = [
    (0, 0),
    (2, 3),
    (5, 4),
    (6, 1),
    (3, 0),
    (1, 5)
]

best_route, best_distance = pso_tsp(cities)

print("\nBest Route:")

for city in best_route:
    print("City", city + 1, end=" -> ")

print("City", best_route[0] + 1)

print("Minimum Distance:", round(best_distance, 2))
```
